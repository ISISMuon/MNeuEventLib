# Getting Started with MNeuEventLib

This tutorial will walk you through the basics of using MNeuEventLib. We will create a data object, apply a filter to it, and generate and save the resulting histogram.

## The `Data` object
The interface of MNeuEventLib centres around the `Data` object, which contains the dataset, filters, and results of a calculation.

The `Data` object takes two mandatory arguments:
- `filename`: the path to the dataset you will analyse.
- `n_spec`: the number of detectors in the instrument used to record the data.

The data we will use in this tutorial is edited data from HIFI, which has 64 detectors. This is an incredibly short run taking place over just three seconds.

````python
from MNeuEventLib import Data

data = Data("../_static/HIFI00195790.nxs", n_spec=64)
````

## Calculating and saving a histogram
To create a histogram output file from an event file, we use the `calculate` method. This function takes your data as well as any filters (see below) and creates a histogram from the event data.

````python
data.calculate()
````

Then, we can save the data using the `save` method. This creates a histogram NeXus file containing the metadata from the event file as well as the calculated histogram.

````python
data.save("output_file.nxs")
````

## Sample log filtering
Converting our event data to a histogram directly is good, but where the event data format really shines is its ability to let us filter out parts of the data that we do not see as relevant. MNeuEventLib allows multiple kinds of filter:
- Filters over **time ranges** ('time filters')
- Filters on the **values of sample logs** ('sample log filters')
- Filters on minimum acceptable **amplitudes** ('amplitude filters')

Here we will look at sample log filters. First, we will look at a sample log taken on this dataset, namely the temperature of the sample over time. We can view a plot of the sample log by importing and using the `plot_sample_log` function from the MNeuEventLib plotting module.

`plot_sample_log` takes two arguments:
- `data`: the Data object to get the sample log from;
- `log_name`: the name of the sample log to plot.

````python
from MNeuEventLib.plotting import plot_sample_log

plot_sample_log(data, "Temp")
````

We can see that the temperature of the sample increases until around 0.35 seconds in, remains stable around 40 until around 2.7 seconds in, and then drops off. Say we want to ignore all of the data where the temperature is not stable. We can do this by adding a sample log filter.

To add a sample log filter, we use the `add_log_filter` method of the `Data` object. This takes four arguments:
- `name`: A name label for the sample log filter. This must be unique to each sample log filter.
- `log`: The sample log that the filter applies to.
- `min`: The minimum value of the sample log to accept.
- `max`: The maximum value of the sample log to accept.

````python
data.add_log_filter("filter_1", "Temp", 39.7, 40.1)
````

Note that there are also log filter functions for when you want an unbounded range on one side: 
- `add_log_filter_above(name, log, min)` for above;
- `add_log_filter_below(name, log, max)` for below.

For example, we could have instead created our filter using:
```python
data.add_log_filter_above("filter_1", "Temperature", 39.7)
```

Once again, we can then calculate and save our histogram outputs.

````python
data.calculate()
````

We can see from the output here that this filtered data has 50874 events, as opposed to the full data which had 64235.

````python
data.save("output_filtered.nxs")
````
