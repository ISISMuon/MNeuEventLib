# How to batch-process data

There may be cases where you would like to generate data for several sets of filters on the same dataset. This tutorial will lead you through an example of batch-processing the dataset.

This guide assumes you are familiar with filtering in MNeuEventLib. The [filtering how-to guide](../how-to/filtering.ipynb) can teach you how to do this.

## The BatchData interface
For batch-processing data, there is a separate interface for MNeuEventLib called `BatchData`. This interface holds several filter sets and result sets for the same dataset.

In addition to the usual arguments taken by the `Data` object, it also takes an argument `n_filter_sets` which sets how many filter sets you would like to use. For this example, we will use the same dataset as in the [getting started tutorial](./getting_started.ipynb).

````python
from MNeuEventLib import BatchData

data = BatchData("../_static/HIFI00195790.nxs", n_spec=64, n_filter_sets=4)
````

The batch filtering interface is similar to the regular interface, but there is an additional first argument to provide the index of which filter set you'd like to change. This can be a number of a filter set, or can be the string `'all'` to apply the filter to every filter set.

So for example, where the normal interface would have:
```python
data.add_time_filter(name, start, end)
```
the batch interface has:
```python
batch_data.add_time_filter(index, name, start, end)
```
and similar for the other filtering functions. Note that as per usual Python convention, the first dataset is index `0`, _not_ index `1`.

For our example, we will start by filtering our data to remove all data that is below our desired temperature:

````python
data.add_log_filter_above("all", "temperature filter", "Temp", 39.7)
````

and we can see this has added the log filter to all of our batches:

````python
print(data)
````

Now we can start adding individual time filters to our data. Let's say we want to cut our data into four time slices between 0.5 and 2.75 seconds. We can add a time filter to each individual filter set:

````python
data.add_time_filter(0, "time filter", 0.5, 1.0625)
data.add_time_filter(1, "time filter", 1.0625, 1.625)
data.add_time_filter(2, "time filter", 1.625, 2.1875)
data.add_time_filter(3, "time filter", 2.1875, 2.75)
````

Now we can calculate, and see that the results in each set have a different number of events each, as expected.

````python
data.calculate()
print(data)
````

Finally, note that _removing_ filters has the same indexing property as adding them. For example, if we wanted to remove the temperature filter from just filter set `2` we could use

````python
data.remove_log_filter(2, "temperature filter")
````

and despite creating the time filters separately, they all have the same name and therefore can all be removed with the same command:

````python
data.remove_time_filter('all', "time filter")

print(data)
````

If you are using multiple filters as 'slices' along the same lines, we recommend giving them the same name so they can all be removed easily like this.
