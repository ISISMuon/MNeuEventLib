# Plotting

This how-to guide will explore the API plotting functionality available in MNeuEventLib.

First we will import an example data file:

````python
from MNeuEventLib import Data

data = Data("HIFI00195790.nxs", 64)
````

## Sample log plotting

One useful type of plot is plotting the data of a sample log, which may help you figure out where to put a sample log filter. To do this, we import and use the plot_sample_log function.

The function takes two arguments:
```
plot_sample_log(data, log_name)
```
where `data` is your `Data` object, and `log_name` is the name of the sample log you want to plot.

````python
from MNeuEventLib.plotting import plot_sample_log

# plot the sample log "Temp"
plot_sample_log(data, "Temp")
````

## Amplitude histogram plotting

When applying filters to detector amplitudes, it may be useful to see a histogram of the event amplitudes. This can be done with the `plot_amplitudes` function. 

````python
from MNeuEventLib.plotting import plot_amplitudes

plot_amplitudes(data)
````

`plot_amplitudes` also has some optional arguments. You can change the number of bins with the `n_bins` argument:

````python
plot_amplitudes(data, n_bins=30)
````

and you can also specify a max height with the `max_height` argument. Note that any values above the specified max height will be put in the last bin, rather than being ignored.

````python
plot_amplitudes(data, max_height=3000)
````
