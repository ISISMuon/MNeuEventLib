# Plotting

This how-to guide will explore the API plotting functionality available in MNeuEventLib.

First we will import an example data file:

````python
from MNeuEventLib import Data

data = Data("HIFI00195790.nxs", 64)
````

## Sample Log plotting

One useful type of plot is plotting the data of a sample log, which may help you figure out where to put a sample log filter. To do this, we import and use the plot_sample_log function.

The function takes two arguments:
```
plot_sample_log(data, log_name)
```
where `data` is your `Data` object, and `log_name` is the name of the sample log you want to plot.

````python
from MNeuEventLib.plotting import plot_sample_log

# plot the sample log "Temp"
plot_sample_log(data, "Temp")
````
