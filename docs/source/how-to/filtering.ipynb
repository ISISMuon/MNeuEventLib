# How to filter data

This how-to guide will walk through the different types of filtering available in MNeuEventLib and how to use them. We assume you understand the material in the getting started tutorial.

Note that all of these kinds of filters can be combined! 

````python
from MNeuEventLib import Data
````

## Time filtering

The simplest kind of filtering is filtering time ranges from the data. These filters can be set to either include the given time ranges, or exclude the given time ranges. Whether the filters include or exclude data is chosen with the `set_time_type` method of the `Data` object. The default is to include.

````python
data = Data("../_static/HIFI00195790.nxs", n_spec=64)

print(data)  # the default time filter type is include
````

````python
data.set_time_type("exclude")

print(data)  # note it has changed to 'exclude'
````

Now we can add a time filter. This is done with the `add_time_filter` method of `Data`. We need to give three arguments:
- `name`: The name of the filter.
- `start`: The start of the filter, in seconds.
- `end`: The end of the filter, in seconds.

For example, as our time filter type is set to `exclude`, the following filter will exclude times from 0.5 seconds to 1.5 seconds from our data.

````python
data.add_time_filter("filter 1", 0.5, 1.5)

print(data)
````

Now, if we calculate our data, we will see in the histogram section of our data that some events have been removed. The full dataset contains 64235 events.

````python
data.calculate()
````

Now if we change the time filter type to `include` and calculate again, we will exclude all data _except_ that between 0.5 and 1.5 seconds. We'd expect our output histogram to have `64235 - 49115 =` 15120 events, which it does:

````python
data.set_time_type("include")
data.calculate()
````

Finally, we can remove our time filter with the `remove_time_filter` method:

````python
data.remove_time_filter("filter 1")
````

## Sample log filtering

Here we will look at sample log filters. Sample logs are various kinds of data taken during an experiment. We will look at a sample log taken on this dataset, namely the temperature of the sample over time.

To add a sample log filter, we use the `add_log_filter` method of the `Data` object. This takes four arguments:
- `name`: A name label for the sample log filter. This must be unique to each sample log filter.
- `log`: The sample log that the filter applies to.
- `lower`: The minimum value of the sample log to accept.
- `upper`: The maximum value of the sample log to accept.

````python
data = Data("../_static/HIFI00195790.nxs", n_spec=64)
````

````python
data.add_log_filter("temperature filter", "Temp", 39.7, 40.1)

data.calculate()
````

We can see from the output here that this filtered data has 50874 events, as opposed to the full data which had 64235.

Note that there are also log filter functions for when you want an unbounded range on one side: 
- `add_log_filter_above(name, log, lower)` for above;
- `add_log_filter_below(name, log, upper)` for below.

For example, if we didn't care about the upper bound, we could have instead created our filter using:
```python
data.add_log_filter_above("temperature filter", "Temperature", 39.7)
```

Note you can also create multiple filters on the same sample log. This filter will also include any data below 38.5 degrees, so we should see more events added to the histogram.

````python
data.add_log_filter_below("temperature filter 2", "Temp", 38.5)

data.calculate()
````

As for time filters, we can remove sample log filters with `remove_log_filter`.

````python
data.remove_log_filter("temperature filter")
data.remove_log_filter("temperature filter 2")
````

## Amplitude filtering

The data can also be filtered according to minimum amplitude filters. These can be per-detector or set as a baseline across all detectors. To set an amplitude filter for one detector use `set_amp`, which has two arguments:
- `detector`: the detector to set the amplitude for.
- `amp`: The minimum allowed amplitude.

````python
data = Data("../_static/HIFI00195790.nxs", n_spec=64)

data.set_amp(6, 2500)
````

To set a baseline, use `set_amps_baseline`, which just takes the minimum amplitude baseline as an argument.

````python
data.set_amps_baseline(1500)
````

````python
data.calculate()
````
